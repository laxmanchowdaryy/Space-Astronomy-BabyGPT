```python
# Read the cleaned dataset 'Astronomy.txt'
with open("Astronomy.txt", "r", encoding="utf-8") as file:
    text = file.read()
print("Astronomy.txt loaded successfully.")
```

```python
#basic stats
num_characters = len(text)
print("Number of characters:", num_characters)
num_words = len(text.split())
print("Number of words:", num_words)
num_lines = len(text.splitlines())
print("Number of lines:", num_lines)
num_paragraphs = len(text.split("\n\n"))
print("Number of paragraphs:", num_paragraphs)
```

```python
# Creating character vocabulary
the_chars=sorted(list(set(text)))
vocab_size=len(the_chars)
print("Vocabulary size:", vocab_size)
print("Characters in vocabulary:")
print("".join(the_chars))
```

```python
#character-to-integer and integer-to-character mappings
stoi = {ch: i for i, ch in enumerate(the_chars)}
itos = {i: ch for i, ch in enumerate(the_chars)}
print("Character-to-integer mapping created.")
print("Integer-to-character mapping created.")
```

```python
#examples of stoi & nitos
print("stoi examples:")
print(list(stoi.items())[:10])
print("\nitos examples:")
print(list(itos.items())[:10])
```

```python
#Encode data into integer tokens
encode=lambda s: [stoi[c] for c in s]
encoded_text=encode(text)
print("Encoding completed.")
print("Number of tokens:", len(encoded_text))
```

```python
#Decoding tokens back into text for check
decode = lambda l: ''.join([itos[i] for i in l])
decoded_text = decode(encoded_text)
print("Decoding completed.")
print(decoded_text[:5000])
```

```python
#checking if decoded data matches oroginal so that we dont loose data
if decoded_text == text:
    print("Verification successful!")
    print("The decoded text is identical to the original text.")
else:
    print("Verification failed.")
```

```python
#Dataset statistics
print("SPACE AND ASTRONOMY DATASET STATISTICS")
print("Characters:", num_characters)
print("Words:", num_words)
print("Lines:", num_lines)
print("Paragraphs:", num_paragraphs)
print("Tokens:", len(encoded_text))
print("Vocabulary size:", vocab_size)
```

```python
#Save the dataset statistics
with open("Astronomy_statistics.txt", "w", encoding="utf-8") as file:
    file.write("SPACE AND ASTRONOMY DATASET STATISTICS\n")
    file.write("======================================\n")
    file.write(f"Characters: {num_characters}\n")
    file.write(f"Words: {num_words}\n")
    file.write(f"Lines: {num_lines}\n")
    file.write(f"Paragraphs: {num_paragraphs}\n")
    file.write(f"Tokens: {len(encoded_text)}\n")
    file.write(f"Vocabulary size: {vocab_size}\n")
print("Astronomy_statistics.txt saved successfully.")
```
