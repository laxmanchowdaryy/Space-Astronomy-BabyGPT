```python
#Read raw data
with open("Astronomy_raw.txt", "r", encoding="utf-8") as file:
    raw_text = file.read()
print("Raw text loaded successfully.")
print("Characters:", len(raw_text))
print("Words:", len(raw_text.split()))
```

```python
#preview to verify
print(raw_text[:5000])
```

```python
# Normalize spaces and blank lines
import re
# Replace multiple spaces or tabs with one space
clean_text = re.sub(r"[ \t]+", " ", raw_text)
# Remove spaces at the beginning and end of each line
clean_text = "\n".join(line.strip() for line in clean_text.splitlines())
# Replace 3 or more blank lines with 2 blank lines
clean_text = re.sub(r"\n{3,}", "\n\n", clean_text)
print("Whitespace normalization completed.")
print("Characters:", len(clean_text))
print("Words:", len(clean_text.split()))
```

```python
# Split the cleaned text into paragraphs
paragraphs = clean_text.split("\n\n")
print("Total paragraphs:", len(paragraphs))
```

```python
# Remove exact duplicate paragraphs
unique_paragraphs = []
seen = set()
for paragraph in paragraphs:
    paragraph = paragraph.strip()
    if paragraph == "":
        continue
    if paragraph not in seen:
        unique_paragraphs.append(paragraph)
        seen.add(paragraph)
print("Paragraphs before duplicate removal:", len(paragraphs))
print("Paragraphs after duplicate removal:", len(unique_paragraphs))
print("Duplicate paragraphs removed:", len(paragraphs) - len(unique_paragraphs))
```

```python
# Combine the cleaned paragraphs
astronomy_text = "\n\n".join(unique_paragraphs)
print("Cleaned corpus created.")
print("Characters:", len(astronomy_text))
print("Words:", len(astronomy_text.split()))
```

```python
# Preview the cleaned corpus
print(astronomy_text[:5000])
```

```python
# Save the cleaned corpus
with open("Astronomy.txt", "w", encoding="utf-8") as file:
    file.write(astronomy_text)
print("Astronomy.txt saved successfully.")
```
